**Завдання 3.** Для того ж набору даних в змінній `sum_purchases` обчисліть **медіану** кожним з зазначених методів:  

  3.1. без використання жодних бібліотек, користуючись тільки Python.  
  3.2. використовуючи відповідний метод бібліотеки numpy.  

Виведіть результат обчислень в кожному з випадків на екран. Порівняйте результати зі значенням вибіркового середнього. Напишіть коментар, для яких задач ви б використовували вибіркове середнє, а для яких - медіану?

```python
# Сортуємо дані
sorted_data = sorted(sum_purchases)

# Знаходимо медіану
n = len(sorted_data)
if n % 2 == 1:
    # Якщо кількість елементів непарна, беремо середній елемент
    median = sorted_data[n // 2]
else:
    # Якщо кількість елементів парна, беремо середнє значення двох середніх елементів
    median = (sorted_data[n // 2 - 1] + sorted_data[n // 2]) / 2

# Виводимо результат
print(f"Медіана без бібліотек: {median}")
```

```python
import numpy as np

# Знаходимо медіану за допомогою numpy
median_numpy = np.median(sum_purchases)

# Виводимо результат
print(f"Медіана за допомогою numpy: {median_numpy}")
```

*Вибіркове середнє: 667.95, а Медіана: 319.0. Це означає, що у вибірці є екстремально великі значення. Тому для оцінки середньої тенденції чи середнього чеку, я б використовувала медіану. Для остаточного рішення, я б шукала інформацію на скільки типові є отримані екстремальні значення. Якщо це звичайна поведінка, я б їх також враховувала для прогнозу продажів, наприклад.*

**Завдання 4**. Обчислість стандартне відхилення наведеної вибірки в змінній `sum_purchases` кожним з зазначених методів:  

  4.1. без використання жодних бібліотек, користуючись тільки Python.  
  4.2. використовуючи відповідний метод бібліотеки numpy.  

Виведіть результат обчислень в кожному з випадків на екран.

```python
# Обчислюємо вибіркове середнє
mean = sum(sum_purchases) / len(sum_purchases)

# Обчислюємо суму квадратів відхилень від середнього
squared_diff_sum = sum((x - mean) ** 2 for x in sum_purchases)

# Обчислюємо стандартне відхилення
standard_deviation = (squared_diff_sum / (len(sum_purchases) - 1)) ** 0.5

# Виводимо результат
print(f"Стандартне відхилення без бібліотек: {standard_deviation}")
```

```python
# Обчислюємо стандартне відхилення за допомогою numpy
std_deviation_numpy = np.std(sum_purchases, ddof=1)  # ddof=1 для вибіркової варіації

# Виводимо результат
print(f"Стандартне відхилення за допомогою numpy: {std_deviation_numpy}")
```

**Завдання 5**. Ми провели дослідження віку студентів-першокурсників. Виявилось, що
- 10 студентів мають вік 16 років
- 25 студентів - 17 років
- 20 студентів - 18 років
- 2 студенти - 20 років

Обчислість середнє значення для цієї вибірки будь-яким способом - на чистому Python або з numpy - та виведіть результат на екран.

```python
# Дані
ages = [16, 17, 18, 20]
students = [10, 25, 20, 2]

# Обчислюємо середнє значення
weighted_sum = sum(age * count for age, count in zip(ages, students))
total_students = sum(students)
mean_age = weighted_sum / total_students

# Виводимо результат
print(f"Середній вік без бібліотек: {mean_age}")
```

```python
import numpy as np

# Дані
ages = np.array([16, 17, 18, 20])
students = np.array([10, 25, 20, 2])

# Обчислюємо середнє значення за допомогою numpy
mean_age_numpy = np.average(ages, weights=students)

# Виводимо результат
print(f"Середній вік за допомогою numpy: {mean_age_numpy}")
```

**Завдання 6**. Крім даних про суму покупок, ми зібрали ще дані про кількість відвідувань кожного з кожного покупця, які знаходяться в змінній `visits`. Обчисліть коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань кожним з зазначених методів:  

  6.1. без використання жодних бібліотек, користуючись тільки Python.  
  6.2. використовуючи відповідний метод бібліотеки numpy.  
  
Виведіть результат на екран. Проаналізуйте результат.

Виведіть додатково графік розсіювання викликавши код нижче і зробіть заключення, чи варто щось змінити при підрахунку кореляції, аби коефіцієнт Пірсона показував коректну оцінку лінійної залежності?

```python
visits = [ 6,  2, 10,  3,  9,  9,  2,  3,  6,  6, 10,  2,  5, 10,  3,  5,  4,
        7,  4,  2]
```

```python
# Обчислюємо середні значення для кожного з наборів даних
mean_purchases = sum(sum_purchases) / len(sum_purchases)
mean_visits = sum(visits) / len(visits)

# Обчислюємо чисельник і знаменник для коефіцієнта кореляції Пірсона
numerator = sum((p - mean_purchases) * (v - mean_visits) for p, v in zip(sum_purchases, visits))
denominator = (sum((p - mean_purchases) ** 2 for p in sum_purchases) * sum((v - mean_visits) ** 2 for v in visits)) ** 0.5

# Обчислюємо коефіцієнт кореляції Пірсона
pearson_correlation = numerator / denominator

# Виводимо результат
print(f"Коефіцієнт кореляції Пірсона без бібліотек: {pearson_correlation}")
```

```python
# Обчислюємо коефіцієнт кореляції Пірсона за допомогою numpy
pearson_correlation_numpy = np.corrcoef(sum_purchases, visits)[0, 1]

# Виводимо результат
print(f"Коефіцієнт кореляції Пірсона за допомогою numpy: {pearson_correlation_numpy}")
```

**Завдання 7**. В попередньому завданні ми помітили, що викиди нам заважають отримати "правдиве" значення кореляції. Давайте вилучимо записи, які відповідають значенням з викидами в сумі покупок і обчисліть коеф. кореляції. Для цього поставте `threshold`, який дорінює середньому значенню + 2 стандартних вдіхилля суми покупок і всі значення, які більше за нього, будемо вважати викидами (є і інші правила для визначення викидів - з ним и познайомимось пізніше).

Вилучіть значення з обох масивів і обчисліть коеф. кореляції Пірсона з допомогою `numpy` для кількості візитів і суми покупок без записів-викидів. Виведіть результат на екран, побудуйте діаграму розсіювання і проаналізуйте результат: що можемо сказати про залежність в даних?

```python
# Обчислюємо середнє та стандартне відхилення для суми покупок
mean_purchases = np.mean(sum_purchases)
std_purchases = np.std(sum_purchases)

# Визначаємо threshold для викидів
threshold = mean_purchases + 2 * std_purchases

# Вилучаємо викиди з обох масивів
filtered_sum_purchases = [p for p in sum_purchases if p <= threshold]
filtered_visits = [v for p, v in zip(sum_purchases, visits) if p <= threshold]

# Обчислюємо коефіцієнт кореляції Пірсона без викидів за допомогою numpy
pearson_correlation_filtered = np.corrcoef(filtered_sum_purchases, filtered_visits)[0, 1]

# Виводимо результат
print(f"Коефіцієнт кореляції Пірсона без викидів: {pearson_correlation_filtered}")

# Будуємо діаграму розсіювання
plt.scatter(filtered_sum_purchases, filtered_visits)
plt.title('Діаграма розсіювання: сума покупок та кількість відвідувань без викидів')
plt.xlabel('Сума покупок')
plt.ylabel('Кількість відвідувань')
plt.show()
```
